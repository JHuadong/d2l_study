# 2.2 Data pre-processing
In order to use deep learning to solve real-world problems, we often **start by pre-processing raw data, rather than working with data that is already in tensor

# 2.2.1 Read the dataset
First, we will create a synthetic dataset and store it in a CSV (comma-separated values) file: ../data/house_tiny.csv.
Data stored in other formats can also be processed in a similar manner. Below, we will write the dataset to the CSV file line by line.

```python
import os

os.makedirs(os.path.join('..','data'), exist_ok=True)
data_file = os.path.join('..', 'data', 'house_tiny.csv')
with open(data_file, 'w') as f:
    f.write('NumRooms,Alley,Price\n') # column name
    f.write('NA,Pave,127500\n') # Each line represents a data sample
    f.write('2,NA,106000\n')
    f.write('4,NA,178100\n')
    f.write('NA,NA,140000\n')
```

To load the raw dataset from the CSV file we have created, we import the pandas package and call the `read_csv` function. The dataset has four rows and three columns. Each
row describes the number of rooms (‘NumRooms’), the type of alley (‘Alley’) and the house price (‘Price’)

```python
import pandas as pd

data = pd.read_csv(data_file, sep=',')
print(data)
```

## 2.2.2 Handling missing values
Please note that the ‘NaN’ entry represents a missing value. **Typical methods for handling missing data include imputation and deletion, whereby imputation involves substituting a
placeholder value for the missing value, whilst deletion simply ignores the missing value.** Here, we shall consider the imputation method.<br><br>
Using the position index `iloc`, we split `data` into `inputs` and `outputs`, where the former consists of the first two columns of `data` and the latter consists of the last column of `data`.
For missing values in `inputs`, we replace the ‘NaN’ entries with the mean of the corresponding column.

```python
inputs, outputs = data.iloc[:, 0:2], data.iloc[:, 2]
inputs = inputs.fillna(inputs.mean(numeric_only=True))
print(inputs)
```
