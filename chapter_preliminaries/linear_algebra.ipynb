## **2.3.1 Scalar**
**Strictly speaking, a value consisting of a single numerical quantity is called a scalar.** <br><br>
This book employs mathematical notation, in which scalar variables are denoted by ordinary lower-case letters (e.g. x, y and z). **In this book, $\mathbb{R}$ denotes the set of all (continuous) real
scalars.** We shall define the concept of a ‘space’ strictly later on, but for now it is sufficient to remember that the expression x ∈ $\mathbb{R}$ is the formal way of stating that x is a real-valued scalar.
The symbol ∈ is called ‘belongs to’ and denotes ‘being a member of a set’. For example, x, y ∈ {0, 1} can be used to indicate that x and y are numbers whose values can only
be 0 or 1.<br><br>
**A scalar is represented by a tensor with only one element.**

```python
import torch

x = torch.tensor(3.0)
y = torch.tensor(2.0)

x + y, x * y, x / y, x**y
```

**x**, **y** and **z**).<br><br>
**Vectors are represented using one-dimensional tensors.** Generally speaking, tensors can be of any length, depending on the machine’s memory limitations.

```python
x = torch.arange(4)
x
```

### **Length, Dimensions and Shape**
In mathematical notation, if we wish to
**state that a vector x consists of n real-valued scalars, we can represent it as x ∈ R^n. The length of a vector is commonly referred to as its dimension.**<br><br>
As with ordinary Python arrays, **we can access the length of a tensor by calling Python’s built-in len() function**.

```python
len(x)
```

When representing a vector (with only one axis) as a tensor, we can also access the vector’s length via the .shape attribute. **The shape is a
list of elements that specifies the length (dimension) of the tensor along each axis. For a tensor with only one axis, the shape consists of a single element.**

```python
x.shape
```

Please note that the term ‘dimension’ often has different meanings in different contexts. Let us clarify this here: **the dimension of a vector or axis is used to denote the length of the vector or axis, that is, the number of elements in the vector or axis. However, the
dimension of a tensor is used to denote the number of axes it possesses. In this sense, the dimension of a particular axis of a tensor is the length of that axis.**

## **2.3.5 Basic Properties of Tensor Algorithms**
Scalars, vectors, matrices and tensors with any number of axes (the term ‘tensor’ in this subsection refers to algebraic objects) possess a number of practical properties. For example, from the definition of
element-wise operations, it can be seen that **any unary element-wise operation does not alter the shape of its operands. Similarly, given any two tensors of the same
shape, the result of any binary element-wise operation will be a tensor of the same shape**. For example, adding two matrices of the same shape
involves performing element-wise addition on these two matrices.

```python
A = torch.arange(20, dtype=torch.float32).reshape(5, 4)
B = A.clone()
A, A + B
```

**More specifically, the element-wise multiplication of two matrices is known as the Hadamard product (symbol ⊙)**. For a matrix $B$ ∈ $\mathbb{R}^{m×n}$,
where the element at row $i$ and column $j$ is $b_{ij}$. The Hadamard product of matrices $A$ (defined in [(2.3.2)](#232) and $B$ is:<br>
##### **(2.3.4)**
![Hadamard product](../images/picture2_3_4.png)

```python
A * B
```

**Multiplying or adding a scalar to a tensor does not alter the tensor’s shape; each element of the tensor is added to or multiplied by the scalar.**

```python
a = 2
X = torch.arange(24).reshape(2, 3, 4)
a + X, (a * X).shape
```

## **2.3.6 Dimension Reduction**
A useful operation on any tensor is to compute the sum of its elements. In mathematical notation, the summation is represented by a symbol $\sum$.

```python
x = torch.arange(4, dtype=torch.float32)
x, x.sum()
```

We can denote the sum of the elements of a tensor of arbitrary shape. For example, the sum of the elements in matrix A can be written as $\sum_{i=1}^{m}\sum_{j=1}^{n}a_{ij}$.

```python
A.shape, A.sum()
```

**By default, calling the sum function reduces the dimension of a tensor along all axes, turning it into a scalar.** We can also specify along which
axis the tensor should be reduced in dimension through summation. Taking a matrix as an example, to reduce the dimension by summing the elements of all rows (axis 0), you can specify
`axis=0` when calling the function. Since the input matrix is reduced in dimension along axis 0 to produce an output vector, the dimension along axis 0 of the input disappears in the output shape.

```python
A_sum_axis0 = A.sum(dim=0)
A_sum_axis0, A_sum_axis0.shape
```

Specifying `axis=1` reduces the dimensionality by aggregating the elements of all columns (axis 1). Therefore, the dimension of input axis 1 is lost in the output shape.

```python
A_sum_axis1 = A.sum(dim=1)
A_sum_axis1, A_sum_axis1.shape
```

**Summing a matrix row by row and column by column is equivalent to summing all of its elements.**

```python
A.sum(dim=[0, 1])
```

A quantity related to summation is the mean (or average). We calculate the mean by dividing the sum by the total number of elements. In code,
we can call a function to calculate the mean of a tensor of any shape.

```python
A.mean(), A.sum() / A.numel()
```

Similarly, a function that computes the average can also reduce the dimension of a tensor along a specified axis.

```python
A.mean(dim=0), A.sum(dim=0) / A.shape[0]
```

### Non-dimension-reduction summation

Sometimes it can be helpful to keep the number of axes constant when calling a function to calculate a sum or an average.

```python
sum_A = A.sum(dim=1, keepdim=True)
sum_A
```

Since `sum_A` still has two axes after summing each row, we can use broadcasting to divide `A` by `sum_A`.

```python
A / sum_A
```

If we want to compute the cumulative sum of element A along a specific axis—for example, axis=0 (computed by row)—we can call the `cumsum` function. This function does not reduce the dimension of the input tensor along
any axis.

```python
A.cumsum(dim=0)
```

## **2.3.7 Dot Production**
Another of the most fundamental operations is the dot product. Given two vectors **x** and **y** ∈ $\mathbb{R}^{d}$, their
dot product $x^{⊤}$**y** (or ⟨**x**,**y**⟩) is the sum of the element-wise products at corresponding positions: **x**$^⊤$**y** = $\sum^{d}
_{i=1}x_{i}y_{i}$.

```python
y = torch.ones(4, dtype=torch.float32)
x, y, torch.dot(x, y)
```

Note that we can **represent the dot product of two vectors by performing element-wise multiplication and then summing the results**:

```python
torch.sum(x * y)
```

## **2.3.10 Norm**
**Informally speaking, the norm of a vector indicates how large the vector is. The concept of “size” considered here
does not involve dimension, but rather the magnitude of the components.**<br><br>
**In linear algebra, a vector norm is a function $f$ that maps a vector to a scalar. For any given vector $x$, the vector norm must satisfy certain properties. The first
property is: if we scale all elements of the vector by a constant factor $\alpha$, its norm is also scaled by the absolute value of the same constant factor**:
##### (2.3.10)
$$f(\alpha x) = |\alpha|f(x)$$<br><br>
**The second property is the familiar triangle inequality**:<br>
##### (2.3.11)
$$f(x + y) \le f(x) + f(y)$$<br><br>
**The third property, simply put, is that the norm must be nonnegative**:
##### (2.3.12)
$$f(x) \ge 0$$<br><br>
This makes sense. In most cases, the smallest possible size of anything is 0. **The last property states that the norm is at its minimum—0—if and only if
the vector consists entirely of zeros**.
##### (2.3.13)
$$\forall i, [x]_{i} = 0 \Leftrightarrow f(x) = 0$$<br><br>
In fact, the Euclidean distance is an $L_2$ norm: Let x be an n-dimensional vector with elements $x_1$, ..., $x_n$. Its $L2$ norm is the square root of the sum of the squares of the vector’s elements:
##### (2.3.14)
$$||x||_{2} = \sqrt{\sum\limits_{i=1}^{n}x_{i}^{2}}$$<br><br>
In particular, the subscript 2 is often omitted in the $L2$ norm, meaning that ∥x∥ is equivalent to $||x||_{2}$. In code, we can calculate the $L_2$ norm of a vector as follows.

```python
u = torch.tensor([3.0, -4.0])
torch.norm(u)
```

In deep learning, the square of the $L_2$ norm is used more frequently, and **the $L_1$ norm is also commonly encountered; it is defined as the sum of the absolute values of the vector’s elements**:
##### (2.3.15)
$$||x||_{1} = \sum\limits_{i=1}^{n}|x_{i}|$$<br><br>
Compared to the $L_2$ norm, the $L_1$ norm is less sensitive to outliers. To compute the $L_1$ norm, we combine the absolute value function with element-wise summation.

```python
torch.abs(u).sum()
```

Both the $L_2$ norm and the $L_1$ norm are special cases of the more general $L_p$ norm:
##### (2.3.16)
$$||x||_{p} = \left(\sqrt{\sum\limits_{i=1}^{n}|x_{i}^{p}|}\right)^{1/p}$$<br><br>
Similar to the $L_2$ norm of a vector, the Frobenius norm of a matrix $X \in \mathbb{R}^{m \times n}$ is the square root of the sum of the squares of its elements:
##### (2.3.17)
$$||x||_{F} = \sqrt{\sum\limits_{i=1}^{m}\sum\limits_{i=1}^{n}x_{ij}^{2}}$$<br><br>

```python
torch.norm(torch.ones(4, 9))
```

### Norm and Goal
In deep learning, we often seek to solve optimization problems: maximizing the probability assigned to observed data; minimizing the
distance between predictions and actual observations. We represent items (such as words, products, or news articles) as vectors in order to minimize the distance between similar items and maximize the distance between
different items. The objective—perhaps the most important component of a deep learning algorithm (aside from the data)—is typically expressed as a norm.
