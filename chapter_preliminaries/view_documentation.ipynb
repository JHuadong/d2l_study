API documentation, other tutorials, and examples provide a wealth of additional resources beyond this book.
## **2.7.1 Find all functions and classes in the module**
**To find out which functions and classes can be called within a module, you can use the `dir` function.**

```python
import torch

print(dir(torch.distributions))
```

**You can generally ignore functions that begin and end with “\_\_” (double underscores), as they are special objects in Python, as well as functions that begin with a single “\_” (single underscore), which are usually built-in functions**.
## **2.7.2 Find instances of specific functions and classes**
**For more specific instructions on how to use a given function or class, you can call the `help` function.**

```python
help(torch.ones)
```

**In Jupyter Notebook, we can use the `?` command to display documentation in a separate browser window.** For example, the list? command will generate content nearly identical to that produced by the `help(list)` command and display it in a new browser window. Additionally, if we use two question marks, such as `list??`, the Python code implementing that function will be displayed.
## **Summary**
* The official documentation provides a wealth of descriptions and examples beyond what is covered in this book.
* You can view the API documentation by calling the `dir` and `help` functions or by using `?` and `??` in a Jupyter notebook.
