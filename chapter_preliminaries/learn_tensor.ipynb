# Getting Started
First, we import `torch`. Please note that although it is known as PyTorch, the code uses `torch` rather than `pytorch`. Here we use d2l package directly.

```python
from d2l import torch as d2l
```

A tensor represents an array of numerical values, which may have multiple dimensions. A tensor with one axis corresponds to a mathematical vector;
a tensor with two axes corresponds to a mathematical matrix; tensors with more than two axes have no specific mathematical name.<br><br>
First, we can use `arange` to create a row vector `x`. This row vector contains the first 12 integers starting from 0; by default, these are created as integers.It is also possible to specify that they be created as floating-point numbers. Each value in a tensor is referred to as an element. For example, the tensor `x` has 12
elements. Unless otherwise specified, new tensors will be stored in memory and computed using CPU-based calculations.

```python
x = d2l.arange(12)
x
```

You can access the shape of a tensor (its dimensions along each axis) via its `shape` attribute.

```python
x.shape
```

If you simply wish to know the total number of elements in a tensor—that is, the product of all elements in its shape—you can check its size. Since we are dealing with
a vector here, its shape is the same as its size.

```python
x.numel()
```

To change the shape of a tensor without altering the number or values of its elements, you can call the `reshape` function. For example, you can convert the tensor `x` from a row vector of shape
(12,) into a matrix of shape (3,4). This new tensor contains the same values as before the conversion, but it is treated as a matrix with 3 rows and 4 columns
. It is important to emphasise that, although the shape of the tensor has changed, its element values remain unchanged. Note that changing the shape of a tensor
does not alter its size.

```python
X = x.reshape(3, 4)
X
```

We do not need to change the shape by manually specifying each dimension. In other words, if our target shape is (height, width), then once
the width is known, the height is automatically calculated; we do not need to perform the division ourselves. In the example above, to obtain a matrix with 3 rows,
we manually specified that it has 3 rows and 4 columns. Fortunately, we can utilize this automatic dimension calculation feature by using -1. In other words, we can
use `x.reshape(-1, 4)` or `x.reshape(3, -1)` instead of `x.reshape(3, 4)`.<br><br>
Sometimes, we wish to initialize a matrix using all zeros, all ones, other constants, or numbers sampled at random from a specific distribution. We can create
a tensor of shape (2, 3, 4) in which all elements are set to 0. The code is as follows:

```python
d2l.zeros((2, 3, 4))
```

Similarly, we can create a tensor of shape (2, 3, 4) in which all elements are set to 1. The code is as follows:

```python
d2l.ones((2, 3, 4))
```

Sometimes we wish to obtain the values of each element in a tensor by randomly sampling from a specific probability distribution. For example, when we construct arrays to serve
as parameters in a neural network, we usually initialize the parameter values randomly. The following code creates a tensor of shape (3, 4), in which
each element is randomly sampled from a standard Gaussian distribution (normal distribution) with a mean of 0 and a standard deviation of 1.

```python
d2l.randn(3, 4)
```

We can also assign specific values to each element of the desired tensor by providing a Python list (or nested list) containing numerical values. Here,
the outermost list corresponds to axis 0, whilst the inner lists correspond to axis 1.

```python
d2l.tensor([[2, 1, 4, 3], [1, 2, 3, 4], [4, 3, 2, 1]])
```
