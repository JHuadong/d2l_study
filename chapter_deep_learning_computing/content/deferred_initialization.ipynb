# **5.3 Deferred initialisation**
So far, we have overlooked the following tasks that need to be carried out when setting up a network:
  * We have defined the network architecture, but have not specified the input dimensions.
  * When adding a layer, we did not specify the output dimensions of the previous layer.
  * When initialising the parameters, we do not even have enough information to determine how many parameters the model should contain.

After all, deep learning frameworks cannot determine the input dimensions of a network. The trick here lies in the framework’s deferred initialisation, whereby the framework only dynamically infers the size of each layer once the data has passed through the model for the first time.

Having identified the shape of the first layer, the framework processes the second layer, and so on, until all shapes are known. Note that in this case, only the first layer requires deferred initialisation, but the framework still initialises the layers sequentially. Once the shapes of all parameters are known, the framework can initialise the parameters.

## **Summary**
* Delaying initialisation enables the framework to automatically infer parameter shapes, making it easier to modify the model architecture and helping to avoid some common errors.
* We can pass data through the model so that the framework can ultimately initialise the parameters.
