# **Question 5.2.1**
**Using the FancyMLP model defined in Section 5.1, access the parameters of each layer.**

```python
from torch import nn

class NestMLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(20, 64), nn.ReLU(),
                                 nn.Linear(64, 32), nn.ReLU())
        self.linear = nn.Linear(32, 16)

    def forward(self, X):
        return self.linear(self.net(X))

model = NestMLP()

# Parameters for accessing the network layer
print("Visit the parameters of the net layer")
for name, param in model.net.named_parameters():
    print(f"Parameter name: {name}, shape: {param.shape}, Parameter: {param}")

# Parameters for accessing the linear layer
print("Visit the parameters of the linear layers")
for name, param in model.linear.named_parameters():
    print(f"Parameter name: {name}, shape: {param.shape}, Parameter: {param}")
```

**Please refer to the documentation on initialisation modules for details of the various initialisation methods.**<br><br>
* `torch.nn.init.uniform_(tensor, a=0.0, b=1.0)` : Extract the padding input tensor from the uniform distribution $U(a,b)$.

* `torch.nn.init.normal_(tensor, mean=0.0, std=1.0)` : Sampling the input tensor from a normal distribution $N(mean, std2)$.

* `torch.nn.init.constant_(tensor, val)` : Initialise the tensor with a specific value.

* `torch.nn.init.ones_(tensor)` : Fill the input tensor with scalar values of 1.

* `torch.nn.init.zeros_(tensor)` : Fill the input tensor with scalar values of 0.

* `torch.nn.init.eye_(tensor)` : Fill the two-dimensional input tensor with identity matrices.

* `torch.nn.init.xavier_uniform_(tensor, gain=1.0)` : Initialise the input tensor by sampling from the uniform distribution $U(−a, a)$, where the value of $a$ is determined by the following formula:
$$a=gain*\sqrt{\frac{6}{fan_{in}+fan_{out}}}$$

# **Question 5.2.3**
**Construct a multi-layer perceptron with a shared parameter layer and train it. During training, observe the parameters and gradients of each layer in the model.**<br><br>
During training, we printed the parameters and gradients for each layer at every epoch. As can be seen, the parameters and gradients for the `shared_fc` layer are identical, as they share the same parameters.

```python
import torch
import torch.nn as nn
import torch.optim as optim

# the model parameters
input_size = 4
hidden_size = 8
output_size = 4
lr = 0.01
epochs = 2

# Building a multi-layer perceptron with a shared parameter layer
shared_fc = nn.Linear(hidden_size, hidden_size)
MLP = nn.Sequential(
    nn.Linear(input_size, hidden_size), nn.ReLU(),
    shared_fc, nn.ReLU(),
    shared_fc, nn.ReLU(),
    nn.Linear(hidden_size, output_size)
)

# the training data
X = torch.randn(1, input_size)
Y = torch.randn(1, output_size)

# optimization
optimizer = optim.SGD(MLP.parameters(), lr=lr)
# Training model
for epoch in range(epochs):
    # Forward propagation and loss calculation
    Y_pred = MLP(X)
    loss = nn.functional.mse_loss(Y_pred, Y)
    # Backpropagation and gradient updating
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    # Print the parameters and gradients for each layer
    for name, param in MLP.named_parameters():
        print(f"Parameter name: {name}, shape: {param.shape}, Parameter: {param}")
    print('Epoch: {}, Loss: {}'.format(epoch, loss.item()))
```

# **Question 5.2.4**
**Why is sharing parameters a good idea?**
1. Memory efficiency: Shared parameters can reduce the number of parameters that need to be stored in the model, thereby reducing memory usage.

2. Faster convergence: Sharing parameters can make the model more stable and accelerate convergence.

3. Improving generalisation: Shared parameters can help the model better capture commonalities in the data, thereby improving its generalisation ability.

4. Enhancing model interpretability: Sharing parameters can make the model more concise and clear, thereby enhancing its interpretability.
