# **Question 4.7.5**
**Suppose the rendering image is too large for the GPU you currently have.**

##### 1.Please try distributing it across multiple GPUs.
We can partition a neural network onto GPUs in three ways: network parallelism, layer parallelism, and data parallelism. Network parallelism distributes the computations of each network layer across different GPUs. Layer parallelism distributes the computations within each layer across different GPUs; for example, the output units of a fully connected layer are split across different GPUs for computation. Data parallelism splits the data so that the same network is computed on different GPUs, and then aggregates the results from each GPU to update the parameters.

##### 2.Compared to small-batch training, what are the advantages and disadvantages?.
The advantage is that clusters with multiple GPUs can train larger models (as with the first two partitioning methods) and train models faster (as with the third partitioning method); however, the disadvantage is that speed may be limited by inter-node communication constraints.
