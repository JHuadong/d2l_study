# **Question 3.3.1**
**If the total loss over a mini-batch is replaced with the average of the mini-batch losses, how should the learning rate be adjusted?**<br><br>
If we replace the total loss over a mini-batch with the average of the mini-batch losses, we need to multiply the learning rate by the batch size. This is because, when calculating gradients, we use information from all samples in the mini-batch. Therefore, if we replace the total loss over a mini-batch with the average of the mini-batch losses, it is equivalent to dividing the gradient for each sample by the batch size. Consequently, we need to multiply the learning rate by the batch size to maintain the same update step size.

# **Question 3.3.2**
**Replace the original loss with the Huber loss, that is:**
$$
l(y,y')=
\begin{cases}
|y-y'|-\frac{\sigma}{2} & |y-y'|>\sigma \\
\frac{1}{2\sigma}(y-y')^{2} & others.
\end{cases}
$$

```python
import numpy as np
import torch
from torch.utils import data
from d2l import torch as d2l
```

```python
true_w = torch.tensor([2, -3.4])
true_b = 4.2
features, labels = d2l.synthetic_data(true_w, true_b, 1000)
```

```python
def load_array(data_arrays, batch_size, is_train=True):  #@save
    dataset = data.TensorDataset(*data_arrays)
    return data.DataLoader(dataset, batch_size, shuffle=is_train)

batch_size = 10
data_iter = load_array((features, labels), batch_size)
```

```python
from torch import nn

net = nn.Sequential(nn.Linear(2, 1))
```

```python
loss = nn.HuberLoss()
```

```python
trainer = torch.optim.SGD(net.parameters(), lr=0.03)
```

```python
num_epochs = 3
for epoch in range(num_epochs):
    for X, y in data_iter:
        l = loss(net(X) ,y)
        trainer.zero_grad()
        l.backward()
        trainer.step()
    l = loss(net(features), labels)
    print(f'epoch {epoch + 1}, loss {l:f}')
```

```python
w = net[0].weight.data
print('the evalution loss of w: ', true_w - w.reshape(true_w.shape))
b = net[0].bias.data
print('the evalution loss of b: ', true_b - b)
```
