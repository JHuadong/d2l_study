# Question 3.2.1
**What happens if we initialize the weights to zero? Does the algorithm still work?**<br><br>
If the weights are initialized to zero, the output of every neuron will be the same, which means that the parameters learned by each neuron will also be the same. Consequently, each neuron will update the same parameters, ultimately causing all neurons to learn the same features. Therefore, initializing the weights to zero renders the algorithm ineffective. This negates the advantage of neural networks, which is their ability to learn different features.

# Question 3.2.2
**Suppose we are trying to build a model of the relationship between voltage and current. Can automatic differentiation be used to learn the model's parameters?**

```python
import torch

from d2l.torch import linreg

# generate the data
x = torch.randn(100, 1)
y = 30 * x

# define the model
model = torch.nn.Linear(1, 1)

# define the loss function and Optimizer
loss = torch.nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# train the model
for epoch in range(500):
    # Forward Propagation
    y_hat = model(x)

    # get the loss
    l = loss(y_hat, y)

    # Back Propagation
    optimizer.zero_grad()
    l.backward()
    optimizer.step()

print(model.weight)
print(model.bias)
```

```python
import torch
x = torch.randn((2), requires_grad=True)
y = x ** 3

# First-order derivative
dy = torch.autograd.grad(y, x, grad_outputs=torch.ones(x.shape),
                         retain_graph=True, create_graph=True)

# Second-order derivative
dy2 = torch.autograd.grad(dy, x, grad_outputs=torch.ones(x.shape))
```

# Question 3.2.6
**Try using different learning rates and observe how quickly the loss function value decreases.?**

```python
%matplotlib inline
import random
import torch
import numpy as np
from d2l import torch as d2l
```

```python
# 生成数据集
def synthetic_data(w, b, num_examples):  #@save
    """y=Xw+b+noise"""
    X = torch.normal(0, 1, (num_examples, len(w)))
    y = torch.matmul(X, w) + b
    y += torch.normal(0, 0.01, y.shape)
    return X, y.reshape((-1, 1))

true_w = torch.tensor([2, -3.4])
true_b = 4.2
features, labels = synthetic_data(true_w, true_b, 1000)
```

```python
# load the dataset
def data_iter(batch_size, features, labels):
    num_examples = len(features)
    indices = list(range(num_examples))
    random.shuffle(indices)
    for i in range(0, num_examples, batch_size):
        batch_indices = torch.tensor(
            indices[i: min(i + batch_size, num_examples)])
        yield features[batch_indices], labels[batch_indices]
```

```python
# initialize the parameters
w = torch.normal(0, 0.01, size=(2,1), requires_grad=True)
b = torch.zeros(1, requires_grad=True)
```

```python
def linreg(X, w, b):  #@save
    return torch.matmul(X, w) + b
```

```python
def squared_loss(y_hat, y):  #@save
    return (y_hat - y.reshape(y_hat.shape)) ** 2 / 2
```

```python
def sgd(params, lr, batch_size):  #@save
    with torch.no_grad():
        for param in params:
            param -= lr * param.grad / batch_size
            param.grad.zero_()
```

```python
lrs = [0.5, 0.3, 0.1, 0.01]
num_epochs = 10
net = linreg
loss = squared_loss

batch_size = 10
all_lrs = []

for lr in lrs:
    train_lrs = []
    for epoch in range(num_epochs):
        for X, y in data_iter(batch_size, features, labels):
            l = loss(net(X, w, b), y)
            l.sum().backward()
            sgd([w, b], lr, batch_size)
        with torch.no_grad():
            train_l = loss(net(features, w, b), labels)
            train_lrs.append(float(train_l.mean()))
    all_lrs.append(train_lrs)
```

```python
epochs = np.arange(1, num_epochs+1)
d2l.plot(epochs, all_lrs, xlabel='epoch num', ylabel='loss',
         legend=[f'learn rate {lr}' for lr in lrs],
         figsize=(6, 4))
```
