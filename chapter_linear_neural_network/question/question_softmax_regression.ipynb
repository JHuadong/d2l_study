# **Question 3.4.1**

**1.Compute the second derivative of the softmax cross-entropy loss $l(y,\hat{y})$.**<br>
$$\partial_{o_j}l(y,\hat{y}) = \frac{exp(o_j)}{\sum_{k=1}^{q}exp(o_k)}-y_j=softmax(o)_j-y_j$$
we can get second derivative then:
$$
\begin{align*}
\partial^{2}_{o_j}l(y,\hat{y}) &= \frac{exp(o_j)*\sum_{k=1}^{q}exp(o_k)-exp(o_j)^2}{\left(\sum_{k=1}^{q}exp(o_k)\right)^2}\\
&= softmax(o)_j - (softmax(o)_j)^2 \\
&= softmax(o)_j(1-softmax(o)_j).
\end{align*}
$$
In particular, softmax(o) is a probability distribution obtained by applying the softmax function to the elements of vector o.

**2.Calculate the variance of the distribution given by softmax(o) and compare it with the second derivative calculated above.<br><br>**
For the softmax function `softmax(o)`, the variance of its distribution is:<br>
$$
\begin{align*}
Var_{softmax(o)} &= \frac{1}{q}\sum\limits_{j=1}^{q}\left(softmax(o)_j - E[softmax(o)_j]\right)^2 \\
&= \frac{1}{q}\sum\limits_{j=1}^{q}\left(softmax(o)_j - \frac{1}{q}\sum\limits_{k=1}^{q}softmax(o)_k\right)^2 \\
&= \frac{1}{q}\sum\limits_{j=1}^{q}\left(softmax(o)_j - \frac{1}{q}\right)^2.
\end{align*}
$$
Expand to:
$$
\begin{align*}
Var_{softmax(o)} &= \frac{1}{q}\sum\limits_{j=1}^{q}\left(softmax(o)_j - \frac{1}{q}\right)^2 \\
&= \frac{1}{q}\left[\left(softmax(o)_1 - \frac{1}{q}\right)^2 + \left(softmax(o)_2 - \frac{1}{q}\right)^2 + ... + \left(softmax(o)_q - \frac{1}{q}\right)^2 \right] \\
&= \frac{1}{q}\left(\sum\limits_{j=1}^{q}softmax^2(o)_j - \frac{2}{q}\sum\limits_{j=1}^{q}softmax(o)_j + \sum\limits_{j=1}^{q}\frac{1}{q^2}\right) \\
&= \frac{1}{q}\left(\sum\limits_{j=1}^{q}softmax^2(o)_j - \frac{2}{q} + \frac{1}{q}\right) \\
&= \frac{1}{q}\sum\limits_{j=1}^{q}softmax^2(o)_j - \frac{1}{q^2}.\\
\end{align*}
$$
The second-order derivative corresponds to:
$$
\begin{align*}
Var(o) &= \frac{1}{q}\sum\limits_{j=1}^{q}softmax^2(o)_j - \frac{1}{q^2} \\
&= -\frac{1}{q}\left(1 - \sum\limits_{j=1}^{q}softmax^2(o)_j\right) + \frac{1}{q} - \frac{1}{q^2} \\
&= -\frac{1}{q}\left(\sum\limits_{j=1}^{q}softmax(o)_j - \sum\limits_{j=1}^{q}softmax^2(o)_j\right) + \frac{1}{q} - \frac{1}{q^2} \\
&= -\frac{1}{q}\sum\limits_{j=1}^{q}\left(softmax(o)_j - softmax^2(o)_j\right) + \frac{1}{q} - \frac{1}{q^2} \\
&= -\frac{1}{q}\sum\limits_{j=1}^{q}\partial^{2}_{o_j}l(y,\hat{y}) - \frac{q-1}{q^2}.\\
\end{align*}
$$

# **Question 3.4.2**
Suppose there are three categories: $A$, $B$, and $C$, each with an equal probability of 1/3.<br><br>


**1.If we try to design binary code for it, what problems might arise?**<br><br>
$$H(X)=log_{2}3\approx1.585 bit$$
This is the theoretical lower bound for encoding each observation.<br><br>
The length of a binary codeword must be an integer. It is not possible for all three symbols to use 1 bit, because there are only two codewords for 1 bit. The optimal single-symbol binary prefix code is the Huffman code:
$$A \rightarrow 0, B \rightarrow 10, C \rightarrow 11$$
The average code length is
$$L_{1} = \frac{1}{3} * 1 + \frac{1}{3} * 2 + \frac{1}{3} * 2=\frac{5}{3}\approx1.667$$
**So the question is: Symbol-by-symbol binary encoding is limited by the integer code length and cannot achieve an entropy of 1.585 bits.**

**2.Please design a better coding scheme. Hint: What would happen if we tried to code two independent observations? What if we coded n observations jointly?**<br><br>
More generally, combine n independent observations into a “super-symbol.” There are a total of
$$3^n$$
sequence with equal probabilities, whose entropy is
$$H(X^n)=nlog_23$$
Perform joint binary coding on these 3n sequences, such as Huffman coding, arithmetic coding, or block coding. The average code length Ln of any binary prefix code satisfies
$$nlog_23 \le L_n \le nlog_23+1$$
Therefore, the average code length per observation is
$$log_23 \le \frac{L_n}{n} \le log_23+\frac{1}{n}$$
when $n \to \infty$
$$\frac{L_n}{n} \to log_23\approx 1.585$$
**In other words, the longer the block in joint coding, the closer the average number of bits required per observation is to the lower bound of Shannon entropy.**

# **Question 3.4.3**
The true softmax is defined as RealSoftMax(a,b)=log(exp(a)+exp(b)).

**1. Proof RealSoftMax(a,b)>max(a,b).**<br><br>
Based on the problem statement, we know that:
$$RealSoftMax(a,b)=log(exp(a)+exp(b))$$
Based on the properties of the logarithmic function:
$$
\begin{align}
log(exp(a)+exp(b))>log(exp(a))=a \\
log(exp(a)+exp(b))>log(exp(b))=b
\end{align}
$$
Combining the two equations above, we obtain:
$$RealSoftMax(a,b)>max(a,b)$$

**2. Proof  $\lambda^{-1}RealSoftMax(\lambda a \lambda b) > max(a,b)$ holds if and only if $\lambda > 0$**.
$$
\begin{align}
\lambda^{-1}RealSoftMax(\lambda a,\lambda b) &= \lambda^{-1}log(exp(\lambda a) + exp(\lambda b)) \\
&> \lambda^{-1}log(max(\lambda a),exp(\lambda b)) \\
&= log(max(e^{\lambda a \lambda^{-1}},e^{\lambda b \lambda^{-1}})) \\
&= log(max(e^{a},e^{b})) \\
&= max(loge^a,loge^b) \\
&= max(a,b)
\end{align}
$$

**3. Proof as for $\lambda \to \infty$, $\lambda^{-1}RealSoftMax(\lambda a \lambda b) \to max(a,b)$.**<br><br>
Based on the problem statement, we know that:
$$
\begin{align}
f(a,b,\lambda) &= \lambda^{-1}RealSoftMax(\lambda a, \lambda b) \\
&= \lambda^{-1} log(exp(\lambda a) + exp(\lambda b))
\end{align}
$$
then:
$$
\begin{align}
f(a,b,\lambda) &> \lambda^{-1} log(exp(\lambda \cdot max(a,b)) \\
&= max(a,b)
\end{align}
$$
It can also be seen that:
$$
\begin{align}
f(a,b,\lambda) &< \lambda^{-1} 2log(exp(\lambda \cdot max(a,b)) \\
&= \frac{log2}{\lambda} + max(a,b)
\end{align}
$$
when $\lambda \to \infty$
$$\lim\limits_{\lambda \to \infty}\frac{log2}{\lambda} + max(a,b) = max(a,b)$$
By the Squeeze Theorem, we obtain
$$\lim\limits_{\lambda \to \infty}f(a,b,\lambda)=max(a,b)$$

**4. What will softmin look like?**
The formula for the Softmin function is as follows:
$$softmin(x_i)=\frac{exp(-x_i)}{\sum_{j}exp(-x_j)}$$
Similar to the softmax function, the softmin function can also be used for multi-class classification problems. **The difference is that, while the softmax function increases the output probability as the elements in the input tensor become larger, the softmin function decreases the output probability.**
